• Does winning the toss increase the probability of winning the match?

```python
# Check whether toss winner also won the match
df['toss_match_winner'] = df['toss_winner'] == df['winner']

# Count wins and losses
result = df['toss_match_winner'].value_counts()

print(result)
```

• Which team has the highest toss-winning percentage?

```python
# Total tosses won by each team
toss_wins = df['toss_winner'].value_counts()

# Total matches played by each team
matches_played = pd.concat([
    df['team1'],
    df['team2']
]).value_counts()

# Calculate toss-winning percentage
toss_percentage = (toss_wins / matches_played) * 100

# Sort in descending order
toss_percentage = toss_percentage.sort_values(ascending=False)

print(toss_percentage)
```

• Which toss decision results in more match victories?

```python
# Check toss decision and match result
df['toss_match_winner'] = df['toss_winner'] == df['winner']

# Group by toss decision and count successful toss winners
result = df[df['toss_match_winner'] == True]['toss_decision'].value_counts()

print(result)
```

• What percentage of toss winners also won the match?

```python
# Count matches where toss winner also won the match
toss_match_wins = (df['toss_winner'] == df['winner']).sum()

# Total matches
total_matches = len(df)

# Calculate percentage
percentage = (toss_match_wins / total_matches) * 100

print("Toss winners who won the match:", toss_match_wins)
print("Total matches:", total_matches)
print("Percentage:", round(percentage, 2), "%")
```

• Which season had the highest number of fielding decisions?

```python
# Filter fielding decisions
fielding = df[df['toss_decision'] == 'field']

# Count fielding decisions season-wise
result = fielding.groupby('season').size().sort_values(ascending=False)

print(result)
```

```python
print("Season:", result.idxmax())
print("Fielding decisions:", result.max())
```

# 5. Visualization Practice

Bar chart :- Matches played by each season

```python
df.head()
```

```python
match_count = pd.concat([df["team1"], df["team2"]]).value_counts()

print(match_count)
```

```python
plt.figure(figsize=(10,5), dpi=100)

plt.bar(match_count.index, match_count.values)

plt.ticklabel_format(style='plain', axis='y')

plt.xticks(rotation=90)
plt.title("Matches Played by Each Team")
plt.xlabel("Teams")
plt.ylabel("Number of Matches")

plt.show()
```

Count plot :-Match-winning teams

```python
sns.barplot(x=match_count.index, y=match_count.values)
```

Pie chart :Toss decisions

```python
match_count.plot(kind='pie')
```

Line chart Season-wise match trends

```python

# Count matches in each season
season_matches = df['season'].value_counts().sort_index()

# Create line chart
plt.figure(figsize=(12, 5), dpi=100)

sns.lineplot(
    x=season_matches.index,
    y=season_matches.values,
    marker='o',
    linewidth=2
)

plt.title('Season-wise Match Trends')
plt.xlabel('Season')
plt.ylabel('Number of Matches')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

Heatmap : Team vs. team victories

```python
# Count team victories against opponents
data = df.groupby(['winner', 'team1']).size().unstack(fill_value=0)

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(data, annot=True, fmt='d', cmap='Blues')

plt.title('Team vs Team Victories')
plt.xlabel('Opponent Team')
plt.ylabel('Winning Team')
plt.show()
```

Box plot : Winning margin distribution

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.boxplot(y=df['win_by_runs'])

plt.title('Winning Margin Distribution')
plt.ylabel('Winning Margin (Runs)')

plt.show()
```

Histogram : Distribution of winning margins

```python
sns.histplot(df[df['win_by_runs'] > 0]['win_by_runs'], bins=20, kde=True)
plt.title('Distribution of Winning Margins (By Runs)')
plt.show()
```

Count plot Top Player of the Match winners

```python
plt.figure(figsize=(12, 5))

top_players = df['player_of_match'].value_counts().head(10)

sns.barplot(x=top_players.index, y=top_players.values)

plt.title('Top 10 Player of the Match Winners')
plt.xlabel('Player Name')
plt.ylabel('Number of Awards')
plt.xticks(rotation=45)
plt.show()
```

Stacked bar : Batting first vs. chasing victories

```python
win_data = df.groupby(['winner', 'result']).size().unstack(fill_value=0)

win_data[['runs', 'wickets']].plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Batting First vs Chasing Victories')
plt.xlabel('Team')
plt.ylabel('Number of Wins')
plt.xticks(rotation=45)
plt.legend(title='Victory Type')
plt.show()
```

Scatter plot: Toss decision vs. winning margin

```python
sns.scatterplot(
    data=df,
    x='toss_decision',
    y='win_by_runs',
    hue='toss_decision'
)

plt.title('Toss Decision vs Winning Margin')
plt.xlabel('Toss Decision')
plt.ylabel('Winning Margin (Runs)')
plt.show()
```
